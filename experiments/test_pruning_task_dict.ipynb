```python
def manually_indices_repeating(num_heads: int, head_dim: int, pruning_indices: torch.Tensor):
    all_indices = []
    for head_num in range(num_heads):
        all_indices.append(
            pruning_indices+head_num*head_dim)
    return torch.cat(all_indices)
```

WARN: k.out doesnt link to q.out

But q.out - link to all modules.

So we need use only attn.q for prune in attn and mlp.up_proj for mlp pruning. Also for local pruning use tp.prune_out_channels, because it doesnt touch residual stream.

```python
# Compare transfer (activation) pruning with structural Torch-Pruning groups.
# Build every group before mutating the model, then reuse exactly those groups.
from lora_transfer_pruning.usecase.local_pruning import LocalPruning
from lora_transfer_pruning.usecase.torch_pruning_group_mapper import TorchPruningGroupMapper
from utils import evaluate_language_model

ATTN_LAYERS = [0, 8, 16]
MLP_LAYERS = [4, 12, 20]
ATTN_LOCAL_IDXS = [2, 6]  # Local head coordinates; LocalPruning closes RoPE pairs.
MLP_OUT_IDXS = [1, 3, 5, 7]

prune_task = {
    **{f"blocks.{layer}.attn.q": (None, ATTN_LOCAL_IDXS) for layer in ATTN_LAYERS},
    **{f"blocks.{layer}.mlp.up_proj": (None, MLP_OUT_IDXS) for layer in MLP_LAYERS},
}
print("prune_task:")
for module_name, task in prune_task.items():
    print(f"  {module_name}: {task}")

bridge.reset_hooks()
from lora_transfer_pruning.adapter.torch_pruning_group_builder import TorchPruningGroupBuilder
local_pruning = LocalPruning(
    bridge,
    evaluation_blocks[:1].to(DEVICE),
)
pruning_groups = local_pruning.get_torch_pruning_groups(prune_task)
assert len(pruning_groups) == len(prune_task)
print("groups:", len(pruning_groups), [len(group) for group in pruning_groups])

baseline_metrics = evaluate_language_model(
    bridge, evaluation_blocks, batch_size=EVAL_BATCH_SIZE
)

# Transfer pruning: map the structural groups to zeroing hooks.
for group in pruning_groups:
    TorchPruningGroupMapper.prepare_group_for_transfer_pruning(
        bridge, group, rescale=False
    )
transfer_metrics = evaluate_language_model(
    bridge, evaluation_blocks, batch_size=EVAL_BATCH_SIZE
)

# Remove activation hooks before applying the same groups structurally.
bridge.reset_hooks()
for group in pruning_groups:
    group.prune()

# The groups retain DependencyGraph's autograd trace; release it before eval.
import gc
del pruning_groups, local_pruning, group
gc.collect()
torch.cuda.empty_cache()

# Structural attention pruning changes D per head. Compress layer-local RoPE
# tensors and update only the pruned HF attention modules (not global config).
original_head_dim = int(bridge.model.config.head_dim)
local_rope_idxs = TorchPruningGroupBuilder.close_rope_pairs(
    torch.tensor(ATTN_LOCAL_IDXS), original_head_dim
)
keep_rope_idxs = torch.tensor(
    [i for i in range(original_head_dim) if i not in set(local_rope_idxs.tolist())],
    dtype=torch.long,
)
new_head_dim = original_head_dim - len(local_rope_idxs)

def compact_rope_hook(activation, hook):
    return activation.index_select(-1, keep_rope_idxs.to(activation.device))

for layer in ATTN_LAYERS:
    attn = bridge.blocks[layer].attn
    attn._original_component.head_dim = new_head_dim
    attn.hook_cos.add_hook(compact_rope_hook)
    attn.hook_sin.add_hook(compact_rope_hook)

structural_shapes = {}
for layer in ATTN_LAYERS:
    attn = bridge.blocks[layer].attn
    structural_shapes[f"blocks.{layer}.attn"] = {
        "q": tuple(attn.q._original_component.weight.shape),
        "k": tuple(attn.k._original_component.weight.shape),
        "v": tuple(attn.v._original_component.weight.shape),
        "o": tuple(attn.o._original_component.weight.shape),
        "head_dim": attn._original_component.head_dim,
    }
for layer in MLP_LAYERS:
    mlp = bridge.blocks[layer].mlp
    structural_shapes[f"blocks.{layer}.mlp"] = {
        "up": tuple(mlp.up_proj._original_component.weight.shape),
        "gate": tuple(mlp.gate_proj._original_component.weight.shape),
        "out": tuple(mlp.out._original_component.weight.shape),
    }
print("structural shapes:", structural_shapes)

structural_metrics = evaluate_language_model(
    bridge, evaluation_blocks, batch_size=EVAL_BATCH_SIZE
)
comparison = {
    "baseline": baseline_metrics,
    "transfer_activation": transfer_metrics,
    "torch_pruning_structural": structural_metrics,
    "structural_minus_transfer": {
        key: structural_metrics[key] - transfer_metrics[key]
        for key in ("loss", "perplexity")
    },
}
comparison
```
